# Bar Graph

### code

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t
from matplotlib.container import BarContainer

def plot_subject_metrics_with_ci(metrics, colors):
    def ci95_halfwidth(x):
        x = np.asarray(x, dtype=float)
        n = len(x)
        return t.ppf(0.975, df=n - 1) * np.std(x, ddof=1) / np.sqrt(n)

    # Prevent original function from showing the figure before CI bars are added
    original_show = plt.show
    plt.show = lambda *args, **kwargs: None

    try:
        df = plot_subject_metrics(metrics, color=colors)
    finally:
        plt.show = original_show

    # Get the same figure/axis created by plot_subject_metrics
    fig = plt.gcf()
    ax = fig.axes[0]

    bar_containers = [
        c for c in ax.containers
        if isinstance(c, BarContainer)
    ]

    # Add 95% CI to the mean bar of each metric
    for (metric_name, values), container in zip(metrics.items(), bar_containers):
        mean_bar = container.patches[-1]

        x = mean_bar.get_x() + mean_bar.get_width() / 2
        y = mean_bar.get_height()

        ax.errorbar(
            x,
            y,
            yerr=ci95_halfwidth(values),
            fmt="none",
            ecolor="black",
            elinewidth=1.5,
            capsize=4,
            capthick=1.5,
            zorder=10
        )

    plt.show()

    return df
```

```python
metrics = {
    "AUC": AUC_20,
    "Acc": acc_20 / 100.0,   # scale to 0..1 for the left axis
    "F1": F1_20
}
df = plot_subject_metrics_with_ci(metrics, ["#1f77b4", "#ff7f0e", "#2ca02c"])
```
